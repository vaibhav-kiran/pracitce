```python
import tensorflow as tf
from tensorflow.keras import models, layers
```

```python

training_set = tf.keras.preprocessing.image_dataset_from_directory(
    "/Users/vaibhavkiran/Machine_learning/pracitce/dataset/training_set",
    labels='inferred',
    label_mode='binary',
    class_names=None,
    color_mode='rgb',
    batch_size=32,
    image_size=(32, 32),
    shuffle=True,
    seed=None,
    validation_split=None,
    subset=None,
    interpolation='bilinear',
    follow_links=False,
    crop_to_aspect_ratio=False,
    pad_to_aspect_ratio=False,
    data_format=None,
    verbose=True
)

```

```python
test_set = tf.keras.preprocessing.image_dataset_from_directory(
    "/Users/vaibhavkiran/Machine_learning/pracitce/dataset/test_set",
    labels='inferred',
    label_mode='binary',
    class_names=None,
    color_mode='rgb',
    batch_size=32,
    image_size=(32, 32),
    shuffle=True,
    seed=None,
    validation_split=None,
    subset=None,
    interpolation='bilinear',
    follow_links=False,
    crop_to_aspect_ratio=False,
    pad_to_aspect_ratio=False,
    data_format=None,
    verbose=True
)
```

```python
cnn = models.Sequential()
cnn.add(layers.Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)))
cnn.add(layers.MaxPooling2D((2, 2)))
```

```python
cnn.add(layers.Conv2D(32, (3, 3), activation='relu'))
cnn.add(layers.MaxPooling2D((2, 2)))
```

```python
cnn.add(layers.Flatten())
cnn.add(layers.Dense(1, activation='sigmoid'))
```

```python
cnn.summary()
```

```python
cnn.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
cnn.fit(x = training_set, validation_data = test_set, epochs = 25)
```

```python
import numpy as np

from tensorflow.keras.preprocessing import image

test_image = image.load_img(

    "/Users/vaibhavkiran/Machine_learning/pracitce/dataset/training_set/cats/cat.3.jpg",

    target_size=(32, 32)

)

test_image = image.img_to_array(test_image)

test_image = np.expand_dims(test_image, axis=0)

result = cnn.predict(test_image)
if result[0][0] > 0.5:

    prediction = "dogs"

else:

    prediction = "cats"

print(prediction)


```
